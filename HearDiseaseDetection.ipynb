Data Collection and Processing

```python
#loading the csv data to a pandas DataFrame
heart_data = pd.read_csv('heart.csv')
```

1 --> Defective Heart
0 --> Health Heart

Splitting the fetures and Target

```python
x = heart_data.drop(columns='target',axis=1)
y = heart_data['target']
```

```python
print(x)
```

```python
print(y)
```

Split Data into training and testing data

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

Model Training 


Logistic Regression Model

```python
model = LogisticRegression()
```

```python
#traininig the LogisticRegression model with Training data
model.fit(x_train,y_train)
```

Model Evaluation


Accuracy Score

```python
#accuracy on trainging data
x_train_predict = model.predict(x_train)
training_data_acc = accuracy_score(x_train_predict,y_train)
```

```python
print('Accuracy on Training Data : ',training_data_acc)
```

```python
#accuracy on test data
x_test_predict = model.predict(x_test)
test_data_acc = accuracy_score(x_test_predict,y_test)
```

```python
print('Accuracy on Test Data : ',test_data_acc)
```

Building a Predictve System


```python
input_data = (50,0,1,120,244,0,1,162,0,1.1,2,0,2)
#do Some prcessing (change to inp data to numpy array)
inp_data_as_numpy = np.asarray(input_data)
#reshape the numpy array as we are predicting for only 1 instance
inp_data_reshape = inp_data_as_numpy.reshape(1,-1)

prediction = model.predict(inp_data_reshape)

if(prediction[0] == 0):
    print("Person Does Not Have Hear Disease")
else:
    print("Person Have Hear Disease")
```
